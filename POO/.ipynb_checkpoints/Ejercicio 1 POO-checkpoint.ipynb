```python
class Tienda:
    
    def __init__(self, nombre, direccion, num_empleados, ventas_trimestre, tipo_tienda = "Electrodomésticos", inaugurada = "True"):
        self.nombre = nombre
        self.direccion = direccion
        self.empleados = num_empleados
        self.ventas = ventas_trimestre
        self.tipo = tipo_tienda
        self.abierta = inaugurada

    def caracteristicas(self):
        return "Nombre: " + self.nombre + ", Dirección: " + self.direccion + ", Número empleados: " + str(self.empleados) + ", Ventas Trimestrales: " + str(self.ventas) + ", Tipo tienda: " + self.tipo + ", Inaugurada: " + self.abierta
    
    def buscar_tienda(self, tienda):
      for tienda in tiendas:
        if tienda == tienda:
          print(tienda.nombre, tienda.direccion)
    
    def total_ventas(self):
        return sum(self.ventas)
    
    def media_ventas(self):
        return sum(self.ventas) / len(self.ventas) / self.empleados
    
    def ventas_ultimo_mes(self):
        return self.ventas [-1]
    
    def proyeccion_ventas(self, inversion):
      if inversion < 1000:
        for i in range(len(self.ventas)):
          self.ventas[i] = self.ventas[i] * 1.2
        return self.ventas
      else:
        for i in range(len(self.ventas)):
          self.ventas[i] = self.ventas[i] * 1.5
        return self.ventas
```

```python
class Tienda:
  tipo = "Electrodomésticos"
  abierta = True

  def __init__(self, nombre, direccion, num_empleados, ventas_ultimos_3meses):
    self.nombre = nombre
    self.direccion = direccion
    self.empleados = num_empleados
    self.ventas = ventas_ultimos_3meses
  
  def caracteristicas(self):
    return "Nombre: " + self.nombre + ", Direccion: " + self.direccion + ", Número de empleados: " + str(self.empleados) + ", Ventas últimos 3 meses: " + str(self.ventas)
  
  def total_ventas(self):
    return sum(self.ventas)
  
  def media_ventas(self):
    return self.total_ventas() / len(self.ventas) / self.empleados

  def buscar_tienda(self):
    return "Nombre: " + self.nombre + ", Dirección: " + self.direccion

  def ventas_ultimo_mes(self):
    return self.ventas[-1]
  
  def proyeccion_ventas(self, inversion):
    if inversion < 1000:
      self.ventas = [x * 1.2 in self.ventas for x in self.ventas]
    else:
      self.ventas = [x * 1.5 for x in self.ventas]
    return self.ventas
```

```python
balay = Tienda("Balay", "Paseo de la Castellana 110", 120, [2500, 4000, 1300])
bosch = Tienda ("Bosch", "Avenida de la Diagonal 330", 100, [3000, 1500, 2000])
siemens = Tienda ("Siemens", "Calle Castelló 18", 80, [5000, 4500, 7000])
```

```python
print(balay.caracteristicas())
print(balay.total_ventas())
print(balay.media_ventas())
print(balay.buscar_tienda())
print(balay.ventas_ultimo_mes())
print(balay.proyeccion_ventas(1120))
```
