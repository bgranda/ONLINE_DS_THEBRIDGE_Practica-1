## Ejercicio 9
1. Escribe un programa que dada una variable titulo busque los libros que hay en la librería con ese título. Si no se encuentra debe devolver el mensaje `Ese no lo tengo, ¿mola?`
2. Pruena que el programa con uno de tus libros y con otro que no contenga tu librería.

`````python
libreria = [{'titulo': 'La piramide asesinada', 'autor': 'Christian Jacq', 'idioma original': 'francés', 'año publicacion': 1993}, {'titulo': 'Cocina para todos', 'autor': 'Karlos Arguiñano', 'idioma original': 'español', 'año publicacion': 2025}, {'titulo': 'Book of the dead', 'autor': 'Patricia Cornwell', 'idioma original': 'inglés', 'año publicacion': 2008}, {'titulo': 'El joyero de la reina', 'autor': 'Nieves Herrero', 'idioma original': 'español', 'año publicacion': 2022}, {'titulo': 'Cry Wolf', 'autor': 'Patricia Briggs', 'idioma original': 'ingles', 'año publicacion': 2009}]
`````

`````python
def buscar_libro(titulo):
    """
    Busca un libro por título en la librería.
    Retorna la información del libro si existe, o un mensaje si no se encuentra.
    """
    titulo_normalizado = titulo.strip().lower()

    for libro in libreria:
        if libro['titulo'].strip().lower() == titulo_normalizado:
            return f"Libro encontrado: {libro}"

    return "Ese no lo tengo, ¿mola?"

if __name__ == "__main__":
    print(buscar_libro("Cry Wolf"))
    print(buscar_libro("Crepúsculo"))
`````
