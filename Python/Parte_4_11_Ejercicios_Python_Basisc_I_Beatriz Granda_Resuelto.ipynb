## Ejercicio 4
Razona sin ejecutar código el output que obtendremos de las siguientes sentencias
1. True and True and False
2. not ((True or False) and (True or False))
3. (False or False or False or True) and False
4. not False and True and not True

# Operadores
'and' True si ambos valores comparados son True
'or' True si al menos uno de los valores comparados es True
'not' invierte el valor
1. True and True and False = False
2. not((True or False) and (True or False)) = not (True and True) = False
3. (False or False or False or True) and False = True and False = False
4. not False and True and not True = True and True and False = False
