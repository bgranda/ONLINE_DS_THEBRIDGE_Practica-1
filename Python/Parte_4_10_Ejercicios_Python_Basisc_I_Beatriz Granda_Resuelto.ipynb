## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
import math
grados = float(input("El siguiente ángulo en grados es: "))
radianes = grados * (math.pi/180)
print("equivale a este otro angulo en radianes:", radianes)
```

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

```python
base = int(input("Si la base es: "))
altura = int(input("y la altura es: "))
base >= 0 and altura >= 0
print(f"El área del paralelogramo es: {base * altura}")
```
