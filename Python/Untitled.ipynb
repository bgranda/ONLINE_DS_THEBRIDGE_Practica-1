```python
entrada = input("Introduzca 3 números separados por comas: ")
```

```python
numeros = [int(x.strip()) for x in entrada.split(",")]
print(numeros)
```

```python
max(numeros)
```

```python
min(numeros)
```

```python
sum(numeros)
```

```python
# El numero máximo menos el número mínimo es igual a la suma total dividida por 5
max(numeros) - min(numeros) == sum(numeros)/5
```
